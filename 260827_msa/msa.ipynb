# instalar ambiente de anaconda

```python
!conda create -y -q -n msadiv -c conda-forge -c bioconda \
clustalw mafft infernal hmmer tn93
```

```python
!conda run -n msadiv \
clustalw -infile=streptos.ssu.fna -outfile=streptos.ssu.clustalw.fna \
-output=FASTA
```

```python
!conda run -n msadiv \
mafft --thread 2 streptos.ssu.fna > streptos.ssu.mafft.fna
```

```python
!wget --quiet -O RF00177.cm https://rfam.org/family/RF00177/cm
```

```python
!conda run -n msadiv \
cmalign -o streptos.ssu.RF00177.stk RF00177.cm streptos.ssu.fna
```

```python
!conda run -n msadiv \
cmalign --dnaout --outformat afa \
-o streptos.ssu.RF00177.fna.bkp RF00177.cm streptos.ssu.fna
```

```python
!conda run -n msadiv \
esl-reformat --informat afa --gapsym='-' \
-o streptos.ssu.RF00177.fna afa streptos.ssu.RF00177.fna.bkp
```
